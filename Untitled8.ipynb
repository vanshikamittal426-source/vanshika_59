```python
# ============================================================
# Gym & Fitness Center Membership Analytics
# Libraries used: ONLY pandas and numpy (beginner friendly)
# Run with:  python gym_analysis.py
# ============================================================
import pandas as pd
import numpy as np

pd.set_option("display.width", 150)
pd.set_option("display.max_columns", 30)

def heading(text):
    print("\n" + "=" * 70)
    print(text)
    print("=" * 70)

# ------------------------------------------------------------
# TASK 1: Load and inspect the dataset
# ------------------------------------------------------------
# keep_default_na=False  -> pandas would otherwise treat the text "None"
#                           in Fitness_Class as a missing value, but here it
#                           really means "member did not attend a class".
# The file has two columns both called "Checkin_ID". The first (G0000x) is the
# real check-in ID; the second (M0xxx) is clearly the MEMBER ID, so we rename it.
df = pd.read_csv("/content/gym.csv.xls", keep_default_na=False, na_values=[""])
df.columns = ["Checkin_ID", "Member_ID", "Membership_Tier", "Gender",
              "Checkin_Time", "Workout_Duration_Mins", "Fitness_Class", "Monthly_Fee"]

heading("TASK 1: DATA INSPECTION")
print("Shape (rows, columns):", df.shape)
print("\nFirst 5 rows:")
print(df.head())
print("\nData types:")
print(df.dtypes)

# ------------------------------------------------------------
# TASK 2: Missing values and duplicate Checkin_ID
# ------------------------------------------------------------
heading("TASK 2: MISSING VALUES & DUPLICATES")
print("Missing values per column:")
print(df.isna().sum())
dup_ids = df["Checkin_ID"].duplicated().sum()
print("\nDuplicate Checkin_ID rows (extra copies):", dup_ids)
print("Fully identical duplicate rows:", df.duplicated().sum())
df = df.drop_duplicates(subset="Checkin_ID", keep="first").reset_index(drop=True)
print("Rows after removing duplicates:", len(df))

# ------------------------------------------------------------
# TASK 3: Fill missing Workout_Duration_Mins
# ------------------------------------------------------------
heading("TASK 3: CLEAN MISSING WORKOUT DURATION")
median_duration = df["Workout_Duration_Mins"].median()
n_missing = df["Workout_Duration_Mins"].isna().sum()
df["Duration_Was_Missing"] = df["Workout_Duration_Mins"].isna()
df["Workout_Duration_Mins"] = df["Workout_Duration_Mins"].fillna(median_duration)
print(f"Filled {n_missing} missing durations with the MEDIAN = {median_duration} minutes.")
print("Why median? It is not pulled by extreme sessions (unlike the mean), it keeps")
print("the overall distribution almost unchanged, and we lose no rows.")

# ------------------------------------------------------------
# TASK 4: Datetime, weekday and hour
# ------------------------------------------------------------
df["Checkin_Time"] = pd.to_datetime(df["Checkin_Time"], format="%d-%m-%Y %H:%M")
df["Weekday"] = df["Checkin_Time"].dt.day_name()
df["Hour"] = df["Checkin_Time"].dt.hour
df["Month"] = df["Checkin_Time"].dt.strftime("%Y-%m")
heading("TASK 4: TIME FEATURES")
print(df[["Checkin_Time", "Weekday", "Hour", "Month"]].head())

# ------------------------------------------------------------
# TASK 5: Total check-ins and average duration
# ------------------------------------------------------------
heading("TASK 5: TOTAL CHECK-INS & AVERAGE DURATION")
print("Total check-ins:", len(df))
print("Average workout duration (mins):", round(df["Workout_Duration_Mins"].mean(), 2))

# ------------------------------------------------------------
# TASK 6: Visit frequency by membership tier
# ------------------------------------------------------------
heading("TASK 6: TIER COMPARISON")
tier = df.groupby("Membership_Tier").agg(
    Total_Checkins=("Checkin_ID", "count"),
    Unique_Members=("Member_ID", "nunique"),
    Avg_Duration_Mins=("Workout_Duration_Mins", "mean"),
).round(2)
tier["Checkins_Per_Member"] = (tier["Total_Checkins"] / tier["Unique_Members"]).round(2)
tier["Share_of_Checkins_%"] = (tier["Total_Checkins"] / tier["Total_Checkins"].sum() * 100).round(1)
tier = tier.loc[["Basic", "Gold", "Platinum"]]
print(tier)

# ------------------------------------------------------------
# TASK 7: Peak hours and busiest days
# ------------------------------------------------------------
heading("TASK 7: PEAK HOURS & BUSIEST DAYS")
hour_counts = df["Hour"].value_counts().sort_index()
print("Top 5 busiest hours:")
print(hour_counts.sort_values(ascending=False).head(5).rename("Checkins"))
print("\n5 quietest hours:")
print(hour_counts.sort_values().head(5).rename("Checkins"))

day_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
day_counts = df["Weekday"].value_counts().reindex(day_order)
print("\nCheck-ins by weekday:")
print(day_counts.rename("Checkins"))
print("\nBusiest day:", day_counts.idxmax(), "| Quietest day:", day_counts.idxmin())

# Peak-hour schedule summary: group hours into time blocks
df["Time_Block"] = np.select(
    [df["Hour"] < 9, df["Hour"] < 12, df["Hour"] < 17, df["Hour"] < 21],
    ["Early Morning (5-8)", "Late Morning (9-11)", "Afternoon (12-16)", "Evening (17-20)"],
    default="Night (21-23)")
block_order = ["Early Morning (5-8)", "Late Morning (9-11)", "Afternoon (12-16)",
               "Evening (17-20)", "Night (21-23)"]
schedule = df.groupby("Time_Block").size().reindex(block_order).to_frame("Checkins")
schedule["Hours_In_Block"] = [4, 3, 5, 4, 3]
schedule["Checkins_Per_Hour"] = (schedule["Checkins"] / schedule["Hours_In_Block"]).round(1)
schedule["Load"] = np.where(schedule["Checkins_Per_Hour"] >= schedule["Checkins_Per_Hour"].mean(),
                            "Busy", "Quieter")
print("\nPeak-hour schedule summary:")
print(schedule)

# Weekday x time-block heatmap table
print("\nCheck-ins: weekday vs time block")
heat = pd.crosstab(df["Weekday"], df["Time_Block"]).reindex(index=day_order, columns=block_order)
print(heat)

# ------------------------------------------------------------
# TASK 8: Class popularity
# ------------------------------------------------------------
heading("TASK 8: CLASS POPULARITY (ranked)")
classes = df[df["Fitness_Class"] != "None"]
class_table = classes.groupby("Fitness_Class").agg(
    Total_Attendance=("Checkin_ID", "count"),
    Avg_Duration_Mins=("Workout_Duration_Mins", "mean")).round(1)
class_table = class_table.sort_values("Total_Attendance", ascending=False)
class_table["Rank"] = np.arange(1, len(class_table) + 1)
class_table["Share_%"] = (class_table["Total_Attendance"] / class_table["Total_Attendance"].sum() * 100).round(1)
print(class_table)
print("\nCheck-ins with no class (self-guided workout):", (df["Fitness_Class"] == "None").sum())

# ------------------------------------------------------------
# TASK 9: Duration by gender
# ------------------------------------------------------------
heading("TASK 9: AVERAGE DURATION BY GENDER")
gender = df.groupby("Gender")["Workout_Duration_Mins"].agg(["count", "mean", "median"]).round(2)
print(gender)
diff = gender.loc["Male", "mean"] - gender.loc["Female", "mean"]
print(f"\nMale minus Female average difference: {diff:.2f} minutes")

# ------------------------------------------------------------
# TASK 10: Monthly revenue by tier
# ------------------------------------------------------------
heading("TASK 10: MONTHLY MEMBERSHIP REVENUE BY TIER")
# A member pays ONE fee per month, no matter how many times they visit.
# So we count each (member, month, tier) only once, then add up the fees.
# NOTE: in this dataset Monthly_Fee does not match the tier (e.g. some Basic
# rows show 2499), so we use the Monthly_Fee value exactly as recorded.
billing = df.drop_duplicates(subset=["Member_ID", "Month", "Membership_Tier"])
revenue = billing.pivot_table(index="Month", columns="Membership_Tier",
                              values="Monthly_Fee", aggfunc="sum", fill_value=0)
revenue = revenue[["Basic", "Gold", "Platinum"]]
revenue["Total"] = revenue.sum(axis=1)
print(revenue.tail(12))
print("\nTotal revenue by tier (all months):")
print(revenue[["Basic", "Gold", "Platinum", "Total"]].sum())
print("\nAverage monthly revenue by tier:")
print(revenue[["Basic", "Gold", "Platinum", "Total"]].mean().round(0))

# ------------------------------------------------------------
# TASK 11: Low-engagement members (< 2 visits per month)
# ------------------------------------------------------------
heading("TASK 11: LOW-ENGAGEMENT MEMBERS")
n_months = df["Month"].nunique()
member = df.groupby("Member_ID").agg(
    Total_Visits=("Checkin_ID", "count"),
    Active_Months=("Month", "nunique"))
member["Avg_Visits_Per_Month"] = (member["Total_Visits"] / n_months).round(2)
member["Visits_Per_Active_Month"] = (member["Total_Visits"] / member["Active_Months"]).round(2)
member["Low_Engagement"] = member["Avg_Visits_Per_Month"] < 2
print(f"Months covered by the data: {n_months}")
print("Total members:", len(member))
print("Low-engagement members (< 2 visits/month over the whole period):",
      int(member["Low_Engagement"].sum()))
print("Members with < 2 visits in their ACTIVE months:",
      int((member["Visits_Per_Active_Month"] < 2).sum()))
# Month-by-month view: share of member-months with fewer than 2 visits
mm = df.groupby(["Member_ID", "Month"]).size()
print(f"Member-months with fewer than 2 visits: {(mm < 2).sum()} of {len(mm)} "
      f"({(mm < 2).mean()*100:.1f}%)")
# Extra: the literal rule flags everyone, so add two more useful signals
last_date = df["Checkin_Time"].max()
member["Last_Visit"] = df.groupby("Member_ID")["Checkin_Time"].max()
member["Days_Since_Last_Visit"] = (last_date - member["Last_Visit"]).dt.days
member["Single_Visit_Member"] = member["Total_Visits"] == 1
member["Inactive_90_Days"] = member["Days_Since_Last_Visit"] > 90
print("\nExtra signals to prioritise outreach:")
print("Members with only 1 visit ever:", int(member["Single_Visit_Member"].sum()))
print("Members with no visit in the last 90 days:", int(member["Inactive_90_Days"].sum()))
print("Both (1 visit AND inactive 90+ days):",
      int((member["Single_Visit_Member"] & member["Inactive_90_Days"]).sum()))
print("\nHighest-visit members (top 5):")
print(member.sort_values("Total_Visits", ascending=False).head(5))
print("\nVisits per member (how many members have N visits):")
print(member["Total_Visits"].value_counts().sort_index())

# ------------------------------------------------------------
# TASK 12: NumPy statistics
# ------------------------------------------------------------
heading("TASK 12: NUMPY STATISTICS ON WORKOUT DURATION")
d = df["Workout_Duration_Mins"].to_numpy()
q1, q2, q3 = np.percentile(d, [25, 50, 75])
print("Median   :", np.median(d))
print("Variance :", round(np.var(d, ddof=1), 2), "(sample variance)")
print("Std dev  :", round(np.std(d, ddof=1), 2))
print("Q1 (25%) :", q1)
print("Q2 (50%) :", q2)
print("Q3 (75%) :", q3)

# ------------------------------------------------------------
# TASK 13: IQR outliers
# ------------------------------------------------------------
heading("TASK 13: UNUSUAL SESSIONS USING IQR")
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr
df["Duration_Outlier"] = np.where(d < lower, "Unusually Short",
                           np.where(d > upper, "Unusually Long", "Normal"))
print(f"IQR = {iqr} | Lower fence = {lower} | Upper fence = {upper}")
print(df["Duration_Outlier"].value_counts())
print("Shortest session:", d.min(), "mins | Longest session:", d.max(), "mins")

# ------------------------------------------------------------
# TASK 14: Workout intensity label
# ------------------------------------------------------------
heading("TASK 14: WORKOUT INTENSITY LABEL")
# Thresholds (documented): Short < 45 mins | Moderate 45-90 mins | Extended > 90 mins
# (45 and 90 are easy-to-explain round numbers close to Q1=46 and Q3=99.)
df["Workout_Intensity"] = np.select(
    [df["Workout_Duration_Mins"] < 45, df["Workout_Duration_Mins"] <= 90],
    ["Short", "Moderate"], default="Extended")
print(df["Workout_Intensity"].value_counts().reindex(["Short", "Moderate", "Extended"]))
print("\nIntensity by tier (row %):")
print((pd.crosstab(df["Membership_Tier"], df["Workout_Intensity"], normalize="index") * 100)
      .round(1)[["Short", "Moderate", "Extended"]])

# ------------------------------------------------------------
# SAVE CLEANED DATA
# ------------------------------------------------------------
member.reset_index().drop(columns=["Last_Visit"]).to_csv("member_engagement.csv", index=False)
df = df.merge(member[["Low_Engagement"]], left_on="Member_ID", right_index=True, how="left")
df["Checkin_Time"] = df["Checkin_Time"].dt.strftime("%Y-%m-%d %H:%M")
df.drop(columns=["Time_Block"]).to_csv("gym_cleaned.csv", index=False)
# ------------------------------------------------------------
# TASK 15: Recommendations (based on the results above)
# ------------------------------------------------------------
heading("TASK 15: RECOMMENDATIONS")
recs = [
 "1. STAFF FOR THE TWO REAL PEAKS: 7 AM and 1 PM are the busiest hours (63 check-ins each), "
 "and Saturday (137) is the busiest day. Put extra floor staff and cleaning slots at those times; "
 "schedule maintenance on Friday (101, quietest) and 9-11 AM / 7-8 PM dips.",
 "2. FILL QUIET SLOTS WITH CLASSES: demand is fairly flat across the day, so move or add "
 "Strength and Spin (the top 2 classes, 151 and 143 attendances) into quieter windows and "
 "offer off-peak discounts on Fridays to even out the load.",
 "3. RUN A WIN-BACK CAMPAIGN: every member averages under 2 visits/month, and 99 members "
 "visited only once. Send a welcome/re-engagement offer (free PT session, buddy pass) after "
 "30 days with no visit, starting with single-visit and 90-day-inactive members "
 "(see member_engagement.csv).",
 "4. FIX PLATINUM VALUE: Platinum members visit least (1.16 check-ins per member vs 1.71 Basic). "
 "Add visible perks (priority booking, recovery zone, personal trainer sessions) so premium "
 "members use what they pay for - low use is a cancellation risk.",
 "5. PROMOTE CLASS VARIETY AND FIX DATA: 153 check-ins (18%) are self-guided with no class, and "
 "class popularity is almost equal (19-22%), so keep all 5 classes. Also fix billing data: "
 "Monthly_Fee does not match tier (e.g. Basic rows at 2499) and members appear in several "
 "tiers - clean this in the CRM before using revenue numbers for decisions.",
]
for r in recs:
    print(r + "\n")

print("\nSaved cleaned data to gym_cleaned.csv  (rows:", len(df), ")")
```
